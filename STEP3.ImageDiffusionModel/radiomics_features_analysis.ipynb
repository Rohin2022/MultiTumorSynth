```python
corr = df[tumor_columns].corr()
```

```python
import numpy as np
threshold = 0.95

# Only take upper triangle so A-B and B-A aren't duplicated
mask = np.triu(np.ones(corr.shape), k=1).astype(bool)

high_corr = (
    corr.where(mask)
        .stack()
        .reset_index()
)

high_corr.columns = ["feature_1", "feature_2", "correlation"]

high_corr["abs_correlation"] = high_corr["correlation"].abs()

high_corr = high_corr.sort_values(
    "abs_correlation",
    ascending=False
)

```
