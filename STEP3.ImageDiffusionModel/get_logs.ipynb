```python
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator
import pandas as pd

log_dir = "/groups/bodymaps/Rohin/TumorSynthesis/STEP3.ImageDiffusionModel/checkpoints/ddpm/multi_tumor_train/tumor_diffusion_all_radiomics_500tsteps_UnetCA_128dim_v12/logs"

ea = EventAccumulator(log_dir)
ea.Reload()

print(ea.Tags()["scalars"])

for tag in ea.Tags()["scalars"]:
    events = ea.Scalars(tag)

    df = pd.DataFrame({
        "step": [e.step for e in events],
        "value": [e.value for e in events],
        "wall_time": [e.wall_time for e in events],
    })

    safe_tag = tag.replace("/", "_")
    df.to_csv(f"/groups/bodymaps/Rohin/TumorSynthesis/STEP3.ImageDiffusionModel/checkpoints/ddpm/multi_tumor_train/tumor_diffusion_all_radiomics_500tsteps_UnetCA_128dim_v12/tensorboard_data/{safe_tag}.csv", index=False)
```
